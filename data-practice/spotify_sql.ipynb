### **GROUP BY AGGREGATES** KEYWORD

TOP 10 COUNT TRACKS PER ARTIST

```python
sqldf("""

    SELECT
        artists,
        COUNT(track_name) AS track_qty
    FROM
        df
    GROUP BY
        artists
    ORDER BY
        track_qty DESC
    LIMIT 10
  
""")
```

----

### **WHERE** AND **HAVING** KEYWORDS


TOP 10 COUNT TRACKS PER ARTIST BETWEEN 5 AND 10


WHERE = filters rows before aggregation

HAVING = filters groups after aggregation (requires GROUP BY)

```python
sqldf("""

    SELECT
        artists,
        COUNT(track_name) AS track_qty
    FROM
        df
    GROUP BY
        artists
    HAVING
        track_qty >= 100
    ORDER BY
        track_qty DESC
    LIMIT 10
  
""")
```

----

### **LIKE** KEYWORD

```python
sqldf("""

    SELECT
        artists,
        COUNT(track_name) AS track_qty
    FROM
        df       
    WHERE
        artists LIKE 'A%'
    GROUP BY
        artists         
    LIMIT 10
  
""")
```

```python
sqldf("""

    SELECT
        artists,
        COUNT(track_name) AS track_qty
    FROM
        df       
    WHERE
        artists LIKE '%y'
    GROUP BY
        artists         
    LIMIT 10
  
""")
```
